```python
# ============================================================
# 📷 AUTOMATIC WEBCAM FACE DETECTION
# ============================================================

from ultralytics import YOLO
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode

import cv2
import numpy as np
from PIL import Image
import time


def take_photo(quality=0.8):

    js = Javascript("""
    async function takePhoto(quality) {

        const div = document.createElement('div');
        const video = document.createElement('video');

        video.style.display = 'block';
        video.width = 640;
        video.height = 480;

        const stream =
            await navigator.mediaDevices.getUserMedia({
                video: true,
                audio: false
            });

        document.body.appendChild(div);
        div.appendChild(video);

        video.srcObject = stream;

        await video.play();

        // Allow camera to stabilize
        await new Promise(resolve =>
            setTimeout(resolve, 1000)
        );

        const canvas =
            document.createElement('canvas');

        canvas.width = video.videoWidth;
        canvas.height = video.videoHeight;

        canvas
            .getContext('2d')
            .drawImage(video, 0, 0);

        const image =
            canvas.toDataURL(
                'image/jpeg',
                quality
            );

        // Stop camera
        stream
            .getVideoTracks()[0]
            .stop();

        div.remove();

        return image;
    }
    """)

    display(js)

    data = eval_js(
        f"takePhoto({quality})"
    )

    binary = b64decode(
        data.split(',')[1]
    )

    return np.frombuffer(
        binary,
        dtype=np.uint8
    )


# ============================================================
# 🚀 START AUTOMATIC DETECTION
# ============================================================

print("📷 Starting webcam...")
print("⏳ Please allow camera permission if requested.")

frame = take_photo(0.8)

image = cv2.imdecode(
    frame,
    cv2.IMREAD_COLOR
)

# ============================================================
# 🧠 YOLO DETECTION
# ============================================================

results = model.predict(
    source=image,
    conf=0.25,
    imgsz=640,
    verbose=False
)

# Draw detections
annotated = results[0].plot()

# Convert BGR → RGB
annotated = cv2.cvtColor(
    annotated,
    cv2.COLOR_BGR2RGB
)

# ============================================================
# 📊 RESULT
# ============================================================

display(
    Image.fromarray(annotated)
)

face_count = len(
    results[0].boxes
)

print()
print("=" * 50)
print(f"👤 Faces detected: {face_count}")
print("=" * 50)
```

```python
import os
import glob
import shutil
import random

# Original dataset
IMAGE_DIR = "/content/wider-face/images"
LABEL_DIR = "/content/wider-face/labels"

# New YOLO dataset
OUTPUT_DIR = "/content/yolo_dataset"

# Create folders
for folder in [
    "images/train",
    "images/val",
    "labels/train",
    "labels/val"
]:
    os.makedirs(os.path.join(OUTPUT_DIR, folder), exist_ok=True)

# Get all images
images = glob.glob(os.path.join(IMAGE_DIR, "*.jpg"))

print("Total images:", len(images))

# Shuffle
random.seed(42)
random.shuffle(images)

# 80/20 split
split_index = int(len(images) * 0.8)

train_images = images[:split_index]
val_images = images[split_index:]

print("Training images:", len(train_images))
print("Validation images:", len(val_images))

# Copy files
for image_path in train_images:

    filename = os.path.basename(image_path)
    name = os.path.splitext(filename)[0]

    label_path = os.path.join(LABEL_DIR, name + ".txt")

    shutil.copy(
        image_path,
        os.path.join(OUTPUT_DIR, "images/train", filename)
    )

    shutil.copy(
        label_path,
        os.path.join(OUTPUT_DIR, "labels/train", name + ".txt")
    )


for image_path in val_images:

    filename = os.path.basename(image_path)
    name = os.path.splitext(filename)[0]

    label_path = os.path.join(LABEL_DIR, name + ".txt")

    shutil.copy(
        image_path,
        os.path.join(OUTPUT_DIR, "images/val", filename)
    )

    shutil.copy(
        label_path,
        os.path.join(OUTPUT_DIR, "labels/val", name + ".txt")
    )

print("Dataset preparation completed!")
```

```python
import os
import glob
import shutil
import random

IMAGE_DIR = "/content/wider-face/images"
LABEL_DIR = "/content/wider-face/labels"
OUTPUT_DIR = "/content/yolo_dataset"

# Create folders
for folder in [
    "images/train",
    "images/val",
    "labels/train",
    "labels/val"
]:
    os.makedirs(os.path.join(OUTPUT_DIR, folder), exist_ok=True)

# Get images
images = glob.glob(os.path.join(IMAGE_DIR, "*.jpg"))

print("Total images:", len(images))

# Shuffle consistently
random.seed(42)
random.shuffle(images)

# 80% train, 20% validation
split = int(len(images) * 0.8)

train_images = images[:split]
val_images = images[split:]

print("Train:", len(train_images))
print("Validation:", len(val_images))

# Copy training files
for image_path in train_images:
    filename = os.path.basename(image_path)
    name = os.path.splitext(filename)[0]

    label_path = os.path.join(LABEL_DIR, name + ".txt")

    shutil.copy2(
        image_path,
        os.path.join(OUTPUT_DIR, "images/train", filename)
    )

    shutil.copy2(
        label_path,
        os.path.join(OUTPUT_DIR, "labels/train", name + ".txt")
    )

# Copy validation files
for image_path in val_images:
    filename = os.path.basename(image_path)
    name = os.path.splitext(filename)[0]

    label_path = os.path.join(LABEL_DIR, name + ".txt")

    shutil.copy2(
        image_path,
        os.path.join(OUTPUT_DIR, "images/val", filename)
    )

    shutil.copy2(
        label_path,
        os.path.join(OUTPUT_DIR, "labels/val", name + ".txt")
    )

print("✅ Dataset preparation completed!")
```

```python
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO

# Load your trained YOLO model
model = YOLO('runs/detect/train/weights/best.pt')

# Test image path
image_path = 'path/to/test_classroom_image.jpg'
image = cv2.imread(image_path)
results = model(image_path)[0]

cropped_faces = []

for i, box in enumerate(results.boxes):
    # Extract coordinates (x1, y1, x2, y2)
    x1, y1, x2, y2 = map(int, box.xyxy[0])

    # Crop face region
    face_crop = image[y1:y2, x1:x2]
    cropped_faces.append(face_crop)

    # Save cropped face
    cv2.imwrite(f'face_{i}.jpg', face_crop)

print(f"Extracted {len(cropped_faces)} faces.")
```

```python
from IPython.display import display, Javascript
from google.colab.output import eval_js
from base64 import b64decode
import numpy as np
import cv2

# Start camera + capture frame in ONE JS call
js = Javascript("""
async function takeWebcamFrame() {

    // Create video element
    const video = document.createElement('video');

    video.width = 640;
    video.height = 480;
    video.autoplay = true;
    video.playsInline = true;

    video.style.display = 'block';
    video.style.border = '3px solid black';

    document.body.appendChild(video);

    // Request camera
    const stream = await navigator.mediaDevices.getUserMedia({
        video: true,
        audio: false
    });

    video.srcObject = stream;

    // Wait for camera
    await new Promise(resolve => {
        video.onloadedmetadata = resolve;
    });

    await video.play();

    // Give camera time to produce frame
    await new Promise(resolve => setTimeout(resolve, 1000));

    // Capture frame
    const canvas = document.createElement('canvas');

    canvas.width = video.videoWidth;
    canvas.height = video.videoHeight;

    const ctx = canvas.getContext('2d');

    ctx.drawImage(video, 0, 0);

    const image = canvas.toDataURL('image/jpeg', 0.85);

    // Stop camera
    stream.getTracks().forEach(track => track.stop());

    video.remove();

    return image;
}

takeWebcamFrame();
""")

display(js)

# Receive captured image
data = eval_js("takeWebcamFrame()")

# Convert base64 → OpenCV image
image_bytes = b64decode(data.split(",")[1])

frame = cv2.imdecode(
    np.frombuffer(image_bytes, np.uint8),
    cv2.IMREAD_COLOR
)

print("✅ Webcam frame captured!")
print("Frame size:", frame.shape)

display(
    cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
)
```

```python
def get_face_embedding(image_path):
    img = cv2.imread(image_path)

    if img is None:
        print("❌ Could not read:", image_path)
        return None

    faces = app.get(img)

    if len(faces) == 0:
        print("❌ No face detected:", image_path)
        return None

    if len(faces) > 1:
        print("⚠️ Multiple faces detected:", image_path)
        return None

    embedding = faces[0].embedding
    embedding = embedding / np.linalg.norm(embedding)

    return embedding
```

```python
staff_database = []

for staff_folder in os.listdir(FACULTY_DIR):

    staff_path = os.path.join(FACULTY_DIR, staff_folder)

    if not os.path.isdir(staff_path):
        continue

    parts = staff_folder.split("_", 1)

    staff_id = parts[0]
    name = parts[1] if len(parts) > 1 else staff_id

    embeddings = []

    for image_file in os.listdir(staff_path):

        if not image_file.lower().endswith(
            (".jpg", ".jpeg", ".png")
        ):
            continue

        image_path = os.path.join(staff_path, image_file)

        embedding = get_face_embedding(image_path)

        if embedding is not None:
            embeddings.append(embedding)

    if len(embeddings) == 0:
        print(f"❌ No valid images for {staff_folder}")
        continue

    # Average all embeddings for this staff member
    mean_embedding = np.mean(embeddings, axis=0)

    # Normalize
    mean_embedding = mean_embedding / np.linalg.norm(mean_embedding)

    staff_database.append({
        "staff_id": staff_id,
        "name": name,
        "embedding": mean_embedding,
        "num_images": len(embeddings)
    })

    print(
        f"✅ {staff_id} | {name} | "
        f"{len(embeddings)} valid images"
    )
```

## Similarity Function

```python
def cosine_similarity(embedding1, embedding2):
    embedding1 = embedding1 / np.linalg.norm(embedding1)
    embedding2 = embedding2 / np.linalg.norm(embedding2)

    return np.dot(embedding1, embedding2)
```

### Low light enhancement


```python
def enhance_low_light(image):
    # Convert BGR → LAB
    lab = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)

    l, a, b = cv2.split(lab)

    # CLAHE improves local brightness/contrast
    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )

    l = clahe.apply(l)

    # Merge and convert back
    enhanced = cv2.merge((l, a, b))
    enhanced = cv2.cvtColor(enhanced, cv2.COLOR_LAB2BGR)

    return enhanced
```

## Recognition Function

```python
THRESHOLD = 0.60

def recognize_face(query_embedding, staff_database, threshold=THRESHOLD):

    best_match = None
    best_score = -1

    for staff in staff_database:

        score = cosine_similarity(
            query_embedding,
            staff["embedding"]
        )

        if score > best_score:
            best_score = score
            best_match = staff

    if best_score >= threshold:
        return best_match, best_score

    return None, best_score
```

```python
# ============================================================
# FINAL MULTI-FRAME YOLO + INSIGHTFACE RECOGNITION
# Low-Light Support + Frame Voting
# ============================================================

import cv2
import numpy as np
import time
from collections import Counter
from IPython.display import display


# ============================================================
# SETTINGS
# ============================================================

THRESHOLD = 0.60

NUM_FRAMES = 5

FRAME_DELAY = 0.3

YOLO_CONFIDENCE = 0.40

FACE_MARGIN = 0.25


# ============================================================
# LOW-LIGHT ENHANCEMENT
# ============================================================

def enhance_low_light(image):

    # Convert BGR to LAB
    lab = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2LAB
    )

    l, a, b = cv2.split(lab)

    # Improve local brightness
    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )

    l = clahe.apply(l)

    # Merge channels
    enhanced_lab = cv2.merge(
        (l, a, b)
    )

    # Convert back to BGR
    enhanced = cv2.cvtColor(
        enhanced_lab,
        cv2.COLOR_LAB2BGR
    )

    return enhanced


# ============================================================
# COSINE SIMILARITY
# ============================================================

def cosine_similarity(embedding1, embedding2):

    embedding1 = (
        embedding1 /
        np.linalg.norm(embedding1)
    )

    embedding2 = (
        embedding2 /
        np.linalg.norm(embedding2)
    )

    return np.dot(
        embedding1,
        embedding2
    )


# ============================================================
# FACE RECOGNITION
# ============================================================

def recognize_face(
    query_embedding,
    staff_database,
    threshold=THRESHOLD
):

    best_match = None
    best_score = -1

    for staff in staff_database:

        score = cosine_similarity(
            query_embedding,
            staff["embedding"]
        )

        if score > best_score:

            best_score = score
            best_match = staff

    if best_score >= threshold:

        return best_match, best_score

    return None, best_score


# ============================================================
# PROCESS ONE WEBCAM FRAME
# ============================================================

def process_frame(frame):

    # Decode webcam image
    image = cv2.imdecode(
        frame,
        cv2.IMREAD_COLOR
    )

    if image is None:

        return None, []


    # ========================================================
    # YOLO FACE DETECTION
    # ========================================================

    results = model.predict(
        source=image,
        conf=YOLO_CONFIDENCE,
        verbose=False
    )


    detections = []


    # ========================================================
    # PROCESS DETECTED FACES
    # ========================================================

    for result in results:

        if result.boxes is None:
            continue


        for box in result.boxes:

            # YOLO coordinates
            x1, y1, x2, y2 = map(
                int,
                box.xyxy[0].tolist()
            )


            # =================================================
            # EXPAND FACE BOX
            # =================================================

            width = x2 - x1
            height = y2 - y1

            margin_x = int(
                width * FACE_MARGIN
            )

            margin_y = int(
                height * FACE_MARGIN
            )


            crop_x1 = max(
                0,
                x1 - margin_x
            )

            crop_y1 = max(
                0,
                y1 - margin_y
            )

            crop_x2 = min(
                image.shape[1],
                x2 + margin_x
            )

            crop_y2 = min(
                image.shape[0],
                y2 + margin_y
            )


            # =================================================
            # FACE CROP
            # =================================================

            face_crop = image[
                crop_y1:crop_y2,
                crop_x1:crop_x2
            ]


            if face_crop.size == 0:
                continue


            # =================================================
            # INSIGHTFACE - NORMAL IMAGE
            # =================================================

            faces = app.get(
                face_crop
            )


            # =================================================
            # INSIGHTFACE - LOW LIGHT FALLBACK
            # =================================================

            if len(faces) == 0:

                enhanced_crop = enhance_low_light(
                    face_crop
                )

                faces = app.get(
                    enhanced_crop
                )


            # =================================================
            # RECOGNITION
            # =================================================

            best_match = None
            best_score = -1


            if len(faces) > 0:

                # Select largest detected face
                face = max(
                    faces,
                    key=lambda f:
                    (f.bbox[2] - f.bbox[0]) *
                    (f.bbox[3] - f.bbox[1])
                )


                embedding = face.embedding


                # Normalize embedding
                embedding = (
                    embedding /
                    np.linalg.norm(embedding)
                )


                # Compare with staff database
                best_match, best_score = recognize_face(
                    embedding,
                    staff_database,
                    THRESHOLD
                )


            # =================================================
            # SAVE RESULT
            # =================================================

            detections.append({

                "box": (
                    x1,
                    y1,
                    x2,
                    y2
                ),

                "match": best_match,

                "score": best_score

            })


    return image, detections


# ============================================================
# MULTI-FRAME CAPTURE
# ============================================================

print("=" * 60)
print("STARTING MULTI-FRAME FACE RECOGNITION")
print("=" * 60)

print()
print(
    f"Capturing {NUM_FRAMES} frames..."
)

print(
    "Please look at the camera."
)

print()


all_detections = []

last_image = None


# ============================================================
# CAPTURE 5 FRAMES
# ============================================================

for frame_number in range(NUM_FRAMES):

    print(
        f"Frame "
        f"{frame_number + 1}/{NUM_FRAMES}"
    )


    # Capture webcam image
    frame = take_photo(
        quality=0.8
    )


    # Process frame
    image, detections = process_frame(
        frame
    )


    if image is not None:

        last_image = image


    all_detections.append(
        detections
    )


    # Small delay
    if frame_number < NUM_FRAMES - 1:

        time.sleep(
            FRAME_DELAY
        )


print()
print("Processing recognition results...")


# ============================================================
# COLLECT ALL RECOGNIZED PEOPLE
# ============================================================

recognized_results = []


for frame_index, detections in enumerate(
    all_detections
):

    for detection in detections:

        match = detection["match"]

        score = detection["score"]


        if match is not None:

            recognized_results.append({

                "frame": frame_index + 1,

                "staff_id":
                    match["staff_id"],

                "name":
                    match["name"],

                "score":
                    score

            })


# ============================================================
# NO RECOGNITION
# ============================================================

if len(recognized_results) == 0:

    print()
    print("=" * 60)
    print("❌ UNKNOWN")
    print("=" * 60)
    print(
        "No staff member was confidently recognized."
    )
    print(
        "Try better lighting or move closer to the camera."
    )


# ============================================================
# MULTI-FRAME VOTING
# ============================================================

else:

    # Count recognitions for each staff member
    staff_votes = Counter(
        result["staff_id"]
        for result in recognized_results
    )


    # Person with most votes
    winning_staff_id, vote_count = (
        staff_votes.most_common(1)[0]
    )


    # Get scores belonging to winner
    winning_results = [

        result

        for result in recognized_results

        if result["staff_id"]
        == winning_staff_id

    ]


    # Average similarity
    average_score = np.mean([

        result["score"]

        for result in winning_results

    ])


    # Highest similarity
    highest_score = max([

        result["score"]

        for result in winning_results

    ])


    # Find staff information
    final_match = None


    for staff in staff_database:

        if staff["staff_id"] == winning_staff_id:

            final_match = staff

            break


    # ========================================================
    # FINAL DECISION
    # ========================================================

    # At least 2 frames must agree
    minimum_votes = 2


    if (
        final_match is not None
        and vote_count >= minimum_votes
        and average_score >= THRESHOLD
    ):

        final_name = final_match["name"]
        final_id = final_match["staff_id"]

        print()
        print("=" * 60)
        print("✅ STAFF RECOGNIZED")
        print("=" * 60)

        print(
            f"Staff ID       : {final_id}"
        )

        print(
            f"Name           : {final_name}"
        )

        print(
            f"Frames matched : "
            f"{vote_count}/{NUM_FRAMES}"
        )

        print(
            f"Average score  : "
            f"{average_score:.4f}"
        )

        print(
            f"Best score     : "
            f"{highest_score:.4f}"
        )

        print(
            f"Threshold      : "
            f"{THRESHOLD:.2f}"
        )

        print("=" * 60)


    else:

        final_name = "Unknown"

        print()
        print("=" * 60)
        print("⚠️ UNKNOWN / LOW CONFIDENCE")
        print("=" * 60)

        print(
            f"Best candidate : "
            f"{final_match['name'] if final_match else 'None'}"
        )

        print(
            f"Frames matched : "
            f"{vote_count}/{NUM_FRAMES}"
        )

        print(
            f"Average score  : "
            f"{average_score:.4f}"
        )

        print(
            f"Threshold      : "
            f"{THRESHOLD:.2f}"
        )

        print("=" * 60)


# ============================================================
# DISPLAY LAST FRAME
# ============================================================

if last_image is not None:

    display_image = last_image.copy()


    # Use detections from last frame
    if len(all_detections) > 0:

        last_detections = all_detections[-1]


        for detection in last_detections:

            x1, y1, x2, y2 = (
                detection["box"]
            )

            match = detection["match"]

            score = detection["score"]


            # ----------------------------------------------
            # Label
            # ----------------------------------------------

            if match is not None:

                label = (
                    f"{match['name']} "
                    f"{score:.2f}"
                )

                color = (
                    0,
                    255,
                    0
                )

            else:

                label = (
                    f"Unknown "
                    f"{score:.2f}"
                )

                color = (
                    0,
                    0,
                    255
                )


            # ----------------------------------------------
            # Bounding box
            # ----------------------------------------------

            cv2.rectangle(
                display_image,
                (x1, y1),
                (x2, y2),
                color,
                2
            )


            # ----------------------------------------------
            # Label
            # ----------------------------------------------

            cv2.putText(
                display_image,
                label,
                (
                    x1,
                    max(
                        25,
                        y1 - 10
                    )
                ),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                color,
                2
            )


    # Convert BGR → RGB
    display_image = cv2.cvtColor(
        display_image,
        cv2.COLOR_BGR2RGB
    )


    print()
    print("📷 Final captured frame:")

    display(
        display_image
    )
```
